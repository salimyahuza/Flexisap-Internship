# Rock, Paper, Scissors
A game between the user and the computer.
## Step 1: Define the possible choices

```python
import random

CHOICES = ("rock", "paper", "scissors")

# Each key beats its value
WINS_AGAINST = {
    "rock": "scissors",
    "paper": "rock",
    "scissors": "paper",
}
```

## Step 2: Get the user's choice
Accepts full words or first letters (r, p, s) and repeats until valid.

```python
def get_user_choice():
    shortcuts = {"r": "rock", "p": "paper", "s": "scissors"}
    while True:
        answer = input("Enter rock, paper or scissors (r/p/s): ").strip().lower()
        answer = shortcuts.get(answer, answer)
        if answer in CHOICES:
            return answer
        print("Invalid choice. Please try again.")
```

## Step 3: Generate the computer's random choice

```python
def get_computer_choice():
    return random.choice(CHOICES)
```

## Step 4: Determine the winner
Rock beats scissors, scissors beats paper, paper beats rock.

```python
def determine_winner(user, computer):
    if user == computer:
        return "tie"
    if WINS_AGAINST[user] == computer:
        return "user"
    return "computer"
```

## Step 5: Play the game
Plays rounds, shows the result and the score, and asks to play again.

```python
def play_game():
    scores = {"user": 0, "computer": 0, "tie": 0}
    print("Welcome to Rock, Paper, Scissors!\n")

    while True:
        user = get_user_choice()
        computer = get_computer_choice()
        print(f"You chose: {user} | Computer chose: {computer}")

        result = determine_winner(user, computer)
        scores[result] += 1

        if result == "tie":
            print("It's a tie!")
        elif result == "user":
            print(f"You win! {user.capitalize()} beats {computer}.")
        else:
            print(f"Computer wins! {computer.capitalize()} beats {user}.")

        print(f"Score -> You: {scores['user']} | "
              f"Computer: {scores['computer']} | Ties: {scores['tie']}\n")

        again = input("Play again? (y/n): ").strip().lower()
        if again not in ("y", "yes"):
            print("Thanks for playing. Goodbye!")
            break

play_game()
```
